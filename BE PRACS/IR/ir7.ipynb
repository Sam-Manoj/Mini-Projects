```python
# ============================================================
# INFORMATION RETRIEVAL EVALUATION
# Precision, Recall, F1, Precision@K, AP, MAP, DCG, NDCG
# ============================================================
import math
import re
# ============================================================
# STEP 1: CREATE DOCUMENT COLLECTION
# ============================================================
documents = {
    "D1": """Information retrieval is the process of finding relevant
    information from a collection of documents. Search engines
    use information retrieval techniques to answer user queries.""",
    "D2": """Machine learning allows computers to learn patterns from
    data. Classification algorithms are commonly used in
    machine learning applications.""",
    "D3": """Information retrieval systems search documents and return
    relevant information to users. Document indexing improves
    the speed of information retrieval.""",
    "D4": """Search engines use indexing and ranking techniques to find
    relevant documents. Information retrieval is an important
    part of modern search systems.""",
    "D5": """Database management systems are used to store and organize
    structured data. Databases support efficient data retrieval.""",
    "D6": """Computer networks allow devices to communicate with each
    other. Network protocols provide reliable communication
    between different systems.""",
    "D7": """Information retrieval involves searching a collection of
    documents for information that matches a user's query.
    Search techniques help users find relevant documents."""
}
print("=" * 70)
print("DOCUMENT COLLECTION")
print("=" * 70)
for doc_id, text in documents.items():
    print("\n", doc_id)
    print(text.strip())
# ============================================================
# STEP 2: DEFINE QUERY
# ============================================================
query = "information retrieval"
print("\n" + "=" * 70)
print("SEARCH QUERY")
print("=" * 70)
print("Query:", query)
# ============================================================
# SIMPLE TEXT SEARCH
# ============================================================
def preprocess(text):
    text = text.lower()
    words = re.findall(r'\b[a-zA-Z]+\b', text)
    return words
query_words = preprocess(query)
# ============================================================
# CALCULATE RELEVANCE SCORE
# ============================================================
document_scores = {}
for doc_id, text in documents.items():
    words = preprocess(text)
    score = 0
    for word in query_words:
        score += words.count(word)
    document_scores[doc_id] = score
# ============================================================
# RANK DOCUMENTS
# ============================================================
retrieved_documents = sorted(
    document_scores,
    key=lambda x: document_scores[x],
    reverse=True
)
print("\n" + "=" * 70)
print("RETRIEVED DOCUMENTS")
print("=" * 70)
for doc in retrieved_documents:
    print(doc, "-> Score:", document_scores[doc])
# Use top 6 documents
retrieved_documents = retrieved_documents[:6]
print("\nRetrieved Documents:")
for doc in retrieved_documents:
    print(doc)
# ============================================================
# STEP 2(b): RELEVANT DOCUMENTS
# ============================================================
relevant_documents = ["D1", "D3", "D4", "D7"]
print("\nRelevant Documents:")
for doc in relevant_documents:
    print(doc)
# ============================================================
# STEP 3: PRECISION, RECALL AND F1 SCORE
# ============================================================
retrieved_set = set(retrieved_documents)
relevant_set = set(relevant_documents)
true_positive = len(retrieved_set.intersection(relevant_set))
precision = true_positive / len(retrieved_set)
recall = true_positive / len(relevant_set)
if precision + recall == 0:
    f1_score = 0
else:
    f1_score = 2 * precision * recall / (precision + recall)
# ============================================================
# STEP 4: PRECISION@K
# ============================================================
K = 5
top_k = retrieved_documents[:K]
relevant_at_k = 0
for doc in top_k:
    if doc in relevant_set:
        relevant_at_k += 1
precision_at_k = relevant_at_k / K
# ============================================================
# STEP 5: AVERAGE PRECISION
# ============================================================
relevant_count = 0
precision_sum = 0
for rank, doc in enumerate(retrieved_documents, start=1):
    if doc in relevant_set:
        relevant_count += 1
        precision_at_rank = relevant_count / rank
        precision_sum += precision_at_rank
average_precision = precision_sum / len(relevant_set)
# ============================================================
# STEP 6: MAP
# ============================================================
MAP = average_precision
# ============================================================
# STEP 7: DCG AND NDCG
# ============================================================
relevance_scores = []
for doc in retrieved_documents:
    if doc in relevant_set:
        relevance_scores.append(1)
    else:
        relevance_scores.append(0)
DCG = 0
for rank, relevance in enumerate(relevance_scores, start=1):
    DCG += relevance / math.log2(rank + 1)
ideal_scores = sorted(relevance_scores, reverse=True)
IDCG = 0
for rank, relevance in enumerate(ideal_scores, start=1):
    IDCG += relevance / math.log2(rank + 1)
if IDCG == 0:
    NDCG = 0
else:
    NDCG = DCG / IDCG
# ============================================================
# STEP 8: DISPLAY ALL EVALUATION MEASURES
# ============================================================
print("\n\n" + "=" * 70)
print("FINAL EVALUATION RESULTS")
print("=" * 70)
print("\nRetrieved Documents")
for doc in retrieved_documents:
    print(doc)
print("\nRelevant Documents")
for doc in relevant_documents:
    print(doc)
print("\n------------------------------------------")
print("Precision       = {:.3f}".format(precision))
print("Recall          = {:.3f}".format(recall))
print("F1 Score        = {:.3f}".format(f1_score))
print("Precision@{}     = {:.3f}".format(K, precision_at_k))
print("Average Precision = {:.3f}".format(average_precision))
print("MAP             = {:.3f}".format(MAP))
print("DCG             = {:.3f}".format(DCG))
print("NDCG            = {:.3f}".format(NDCG))
print("=" * 70)
```
