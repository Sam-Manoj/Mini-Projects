```python
documents = [
    {"id": "doc_1", "text": "The quick brown fox jumps over the lazy dog. This is a pattern example."},
    {"id": "doc_2", "text": "A lazy cat sat on the mat. Another pattern can be found here."},
    {"id": "doc_3", "text": "The pattern recognition is an interesting field. It has many applications."},
    {"id": "doc_4", "text": "This document does not contain the word pattern."},
    {"id": "doc_5", "text": "Python is a powerful programming language, known for its readability and extensive libraries."},
    {"id": "doc_6", "text": "Data analysis in Python often involves libraries like pandas and numpy."}
]

print(f"Loaded {len(documents)} documents.")
```

```python
import re

# Get user input
pattern = input("Enter search pattern (e.g., 'pat', 'Python'): ").strip()
search_type = input("Enter search type (exact, prefix, suffix, substring, regex, approximate): ").strip().lower()

print(f"\nSearch Pattern: '{pattern}'")
print(f"Search Type: '{search_type}'")
```

```python
from difflib import get_close_matches

matched_results = []

def highlight_match(text, match_start, match_end):
    """Highlights the matched portion of the text."""
    # Using ANSI escape codes for highlighting in console output
    # For rich display in Colab, one might use HTML/Markdown later.
    return text[:match_start] + "\033[1;31m" + text[match_start:match_end] + "\033[0m" + text[match_end:]


for doc in documents:
    doc_id = doc["id"]
    text = doc["text"]

    if search_type == "exact":
        if pattern == text:
            matched_results.append((doc_id, highlight_match(text, 0, len(text))))

    elif search_type == "prefix":
        for sentence in re.split(r'(?<=[.!?]) +', text):
            if sentence.strip().startswith(pattern):
                match_start = sentence.strip().find(pattern)
                match_end = match_start + len(pattern)
                matched_results.append((doc_id, highlight_match(sentence.strip(), match_start, match_end)))

    elif search_type == "suffix":
        for sentence in re.split(r'(?<=[.!?]) +', text):
            if sentence.strip().endswith(pattern):
                match_start = sentence.strip().rfind(pattern)
                match_end = match_start + len(pattern)
                matched_results.append((doc_id, highlight_match(sentence.strip(), match_start, match_end)))

    elif search_type == "substring":
        for sentence in re.split(r'(?<=[.!?]) +', text):
            if pattern in sentence:
                match_start = sentence.find(pattern)
                match_end = match_start + len(pattern)
                matched_results.append((doc_id, highlight_match(sentence, match_start, match_end)))

    elif search_type == "regex":
        for sentence in re.split(r'(?<=[.!?]) +', text):
            match = re.search(pattern, sentence, re.IGNORECASE)
            if match:
                match_start = match.start()
                match_end = match.end()
                matched_results.append((doc_id, highlight_match(sentence, match_start, match_end)))

    elif search_type == "approximate":
        # For approximate match, we'll look for close matches within the words of each sentence.
        # This is a simplification; a more robust solution might involve tokenizing and comparing.
        words = re.findall(r'\b\w+\b', text.lower())
        close_matches = get_close_matches(pattern.lower(), words, n=1, cutoff=0.8)
        if close_matches:
            # Find the first occurrence of the matched word (case-insensitive) in the original text
            matched_word = close_matches[0]
            match_index = text.lower().find(matched_word)
            if match_index != -1:
                matched_results.append((doc_id, highlight_match(text, match_index, match_index + len(matched_word))))

    # Note: Lexical range matching is typically done on pre-sorted data or using specialized data structures (e.g., tries, B-trees).
    # For this example, we'll omit a direct implementation as it's not a simple string operation like the others.
    # If requested, a more complex implementation involving data structures would be needed.

print("Matching complete.")
```

```python
if matched_results:
    print(f"\nFound {len(matched_results)} matches:")
    for doc_id, sentence_with_match in matched_results:
        print(f"Document ID: {doc_id}\nSentence: {sentence_with_match}\n")
else:
    print("\nNo matches found.")
```
