```python
import re
import difflib
# --- Step 1: Load Dataset ---
documents = [
    {"id": 1, "text": "Machine learning is a key field in AI."},
    {"id": 2, "text": "This course covers machine learning algorithms."},
    {"id": 3, "text": "Learning how a machine works is interesting."},
    {"id": 4, "text": "Machine intelligence and deep learning are related."},
    {"id": 5, "text": "The coastal weather is lovely today."},
    {"id": 6, "text": "Computer computation and computing are essential skills."},
    {"id": 7, "text": "There are many flowers in the garden."},
    {"id": 8, "text": "The problem02 was solved efficiently."},
    {"id": 9, "text": "We are looking for proteins in the sample."}
]
def highlight(text, match_str):
    if not match_str:
        return text
    # Use regex for case-insensitive highlighting using ANSI escape codes
    pattern = re.compile(f"({re.escape(match_str)})", re.IGNORECASE)
    return pattern.sub(r'\033[44;37m\1\033[0m', text)
def text_retrieval_system():
    # --- Step 2: Input Pattern & Search Type ---
    print("Available types: exact, prefix, suffix, substring, regex, approximate, range")
    search_type = input("Enter search type: ").strip().lower()
    pattern = input("Enter pattern: ").strip()
    range_end = None
    if search_type == "range":
        range_end = input("Enter end of range: ").strip()
    print(f"\n--- Search Results for {search_type.upper()} matching '{pattern}' ---\n")
    # --- Step 3: Perform Matching ---
    results_found = False
    for doc in documents:
        text = doc['text']
        words = re.findall(r'\w+', text)
        match_found = False
        matched_segment = ""
        if search_type == "exact":
            if any(w.lower() == pattern.lower() for w in words):
                match_found = True
                matched_segment = pattern
        elif search_type == "prefix":
            match = next((w for w in words if w.lower().startswith(pattern.lower())), None)
            if match:
                match_found = True
                matched_segment = match
        elif search_type == "suffix":
            match = next((w for w in words if w.lower().endswith(pattern.lower())), None)
            if match:
                match_found = True
                matched_segment = match
        elif search_type == "substring":
            if pattern.lower() in text.lower():
                match_found = True
                matched_segment = pattern
        elif search_type == "regex":
            regex_match = re.search(pattern, text, re.IGNORECASE)
            if regex_match:
                match_found = True
                matched_segment = regex_match.group()
        elif search_type == "approximate":
            matches = difflib.get_close_matches(pattern.lower(), [w.lower() for w in words], cutoff=0.6)
            if matches:
                match_found = True
                matched_segment = matches[0]
        elif search_type == "range":
            match = next((w for w in words if pattern.lower() <= w.lower() <= range_end.lower()), None)
            if match:
                match_found = True
                matched_segment = match
        # --- Step 4: Display Results ---
        if match_found:
            results_found = True
            highlighted_text = highlight(text, matched_segment)
            print(f"Document ID: {doc['id']}")
            print(f"Sentence: {highlighted_text}\n")
    if not results_found:
        print("No documents matched the criteria.")
# Run the system
text_retrieval_system()
```
